## Dream 2000 PC & Laptops Accessories  Scrapping

```python
import selenium 
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
from selenium.common.exceptions import NoSuchElementException
import time

## -- Options --
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option("useAutomationExtension", False)
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")


## -- PC_Laptops_Accessories List --
PC_Laptops_Accessories = []

urls = ['https://dream2000.com/en/collections/mouses' , 'https://dream2000.com/en/collections/laptop-bags' , 'https://dream2000.com/en/collections/keyboards' , 'https://dream2000.com/en/collections/headset' , 'https://dream2000.com/en/collections/hub' , 'https://dream2000.com/en/collections/computer-cables' , 'https://dream2000.com/en/collections/printer' , 'https://dream2000.com/en/collections/monitors']

## -- Scraping --

for url in urls :
    
    driver = webdriver.Chrome(options=options)
    driver.get(url)
    time.sleep(5)

    page = 1

    while True:

        page_products = 0

        ## -- Product Cards --
        cards = driver.find_elements(By.CSS_SELECTOR,'.product-card__wrapper.color-scheme-theme.anm-reveal-item.overflow-hidden')

        for card in cards:

            try:
                Product = card.find_element(By.CSS_SELECTOR, 'a[href*="/en/products/"]')

                Name = Product.get_attribute('aria-label')
                Link = Product.get_attribute('href')

                Image = card.find_element(By.CSS_SELECTOR, 'img.product-card__image').get_attribute('src')

            except NoSuchElementException:
                continue

            ## -- Price --
            try:
                Price = card.find_element(By.CSS_SELECTOR, '.price').text

            except NoSuchElementException:
                Price = "Product Not Available"

            ## -- Stock --
            try:
                Stock = card.find_element(By.CSS_SELECTOR,'[data-inventory-status]').get_attribute('data-inventory-status')

            except NoSuchElementException:
                Stock = "Product Not Available"

            PC_Laptops_Accessories.append({
                'Name': Name,
                'Link': Link,
                'Image': Image,
                'Price': Price,
                'Stock': Stock
            })

            page_products += 1

        ## -- Page Summary --

        print(
            f"Page {page} finished successfully. "
            f"Products collected on this page: {page_products}. "
            f"Total products collected: {len(PC_Laptops_Accessories)}"
        )

        ## -- Next Page --
        next_buttons = driver.find_elements(
            By.CSS_SELECTOR, "a[aria-label='Next']"
        )

        if not next_buttons:
            print("No Next page. Scraping finished.")
            break

        next_url = next_buttons[0].get_attribute('href')

        if not next_url or next_url == driver.current_url:
            print("Invalid or unchanged Next URL.")
            break

        print("Navigating to:", next_url)

        driver.get(next_url)

        page += 1
        time.sleep(5)


    ## -- Final Result --
    print(f"\nScraping completed. Total products: {len(PC_Laptops_Accessories)}")
```
